# Bubble Sort

```python
def bubble_sort(arr):
    arr = list(arr)
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr


data = [64, 34, 25, 12, 22, 11, 90]
print("Original:", data)
print("Sorted:  ", bubble_sort(data))
```
